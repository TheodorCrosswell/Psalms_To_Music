```python
import os
import time
import requests
import easyocr
import re
import csv  # Import the built-in CSV module


def main():
    # Configuration
    base_url = "https://hymnary.org/page/fetch/SSSH1989/{}/high"
    start_page = 1
    end_page = 502

    target_delay = 5.0
    output_dir = "hymnal_images"
    output_csv = "hymnal_extracted_text.csv"
    os.makedirs(output_dir, exist_ok=True)

    # --- NEW: Initialize the CSV file with headers before the loop ---
    with open(output_csv, mode="w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=["page_number", "url", "text"])
        writer.writeheader()

    session = requests.Session()
    session.headers.update(
        {
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
            "Accept": "image/avif,image/webp,image/apng,image/svg+xml,image/*,*/*;q=0.8",
            "Accept-Language": "en-US,en;q=0.9",
            "Accept-Encoding": "gzip, deflate, br",
            "Referer": "https://hymnary.org/hymnal/SSSH1989",
            "Connection": "keep-alive",
        }
    )

    print("Visiting the main site to establish a session and grab cookies...")
    try:
        session.get("https://hymnary.org/hymnal/SSSH1989", timeout=15)
    except Exception as e:
        print(
            f"Warning: Failed to reach main page for cookies. Continuing anyway... ({e})"
        )

    print("Loading OCR model into memory (this may take a moment)...")
    reader = easyocr.Reader(["en"], gpu=False)

    print(f"Starting extraction for {end_page - start_page + 1} pages...")

    for page_num in range(start_page, end_page + 1):
        loop_start_time = time.time()
        url = base_url.format(page_num)

        print(f"Processing page {page_num}/{end_page}...")

        # Prepare an empty row dictionary
        row_data = {"page_number": page_num, "url": url, "text": ""}

        try:
            response = session.get(url, timeout=15)
            response.raise_for_status()

            image_path = os.path.join(output_dir, f"page_{page_num}.png")
            with open(image_path, "wb") as f:
                f.write(response.content)

            ocr_data = reader.readtext(image_path, paragraph=True, x_ths=2.5, y_ths=0.2)

            valid_lines = []
            for item in ocr_data:
                raw_text = item[1]
                if re.search(r"[A-Za-z]{2,}", raw_text):
                    valid_lines.append(raw_text)

            joined_text = " [NEWLINE] ".join(valid_lines)
            cleaned_text = re.sub(
                r"([A-Za-z]+)\s*-\s*([A-Za-z]+)", r"\1\2", joined_text
            )

            row_data["text"] = cleaned_text

        except KeyboardInterrupt:
            # Safely exit if you press Ctrl+C
            print("\n[!] Script manually interrupted. Exiting gracefully...")
            break
        except Exception as e:
            print(f"  [!] Failed to process page {page_num}. Error: {e}")
            row_data["text"] = "[EXTRACTION FAILED]"

        # --- NEW: Append the row to the CSV immediately ---
        with open(output_csv, mode="a", newline="", encoding="utf-8") as f:
            writer = csv.DictWriter(f, fieldnames=["page_number", "url", "text"])
            writer.writerow(row_data)

        # Delay calculation
        elapsed_time = time.time() - loop_start_time
        sleep_time = target_delay - elapsed_time

        if sleep_time > 0 and page_num < end_page:
            time.sleep(sleep_time)

    print(
        f"Done! Text saved continuously to {output_csv} and images saved in '{output_dir}'."
    )


if __name__ == "__main__":
    main()
```
